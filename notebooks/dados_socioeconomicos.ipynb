## Agregação em um mesmo arquivo dos dados socioeconomicos adicionais coletados, para uso no treinamento do modelo.

```python
import pandas as pd
import numpy as np
```

```python
df_ideb = pd.read_csv("/content/br_historico_ideb_municipal.csv", delimiter=';')
df_gini = pd.read_csv("/content/br_indice_gini_uf.csv", delimiter=';')
df_pbf_2023 = pd.read_csv("/content/br_proporcao_pbf_2023.csv", delimiter=';')
df_pbf_2024 = pd.read_csv("/content/br_proporcao_pbf_2024.csv", delimiter=';')
```

```python
df_ideb = df_ideb.rename(columns={"nome_municipio": "municipio"})
df_ideb['municipio'] = df_ideb['municipio'].str.upper()
```

```python
df_gini_2023 = df_gini.filter(items=["Sigla", "2023"]).rename(columns={"Sigla": "sigla_uf", "2023": "gini"})
df_gini_2024 = df_gini.filter(items=["Sigla", "2024"]).rename(columns={"Sigla": "sigla_uf", "2024": "gini"})
```

```python
df_gini_2023['gini'] = df_gini_2023['gini'].replace(",", ".", regex=True).astype(np.float64)
df_gini_2024['gini'] = df_gini_2024['gini'].replace(",", ".", regex=True).astype(np.float64)
```

```python
df_pbf_2023 = df_pbf_2023.merge(df_gini_2023, on="sigla_uf", how="left").filter(items=["municipio", "sigla_uf", "proporcao", "gini"]).rename(columns={"proporcao": "proporcao_pbf_municipio", "gini": "gini_uf"})
df_pbf_2024 = df_pbf_2024.merge(df_gini_2024, on="sigla_uf", how="left").filter(items=["municipio", "sigla_uf", "proporcao", "gini"]).rename(columns={"proporcao": "proporcao_pbf_municipio", "gini": "gini_uf"})
```

```python
df_pbf_2023.head()
```

```python
df_pbf_2024.head()
```

```python
df_pbf_2023 = df_pbf_2023.merge(df_ideb, on=["municipio", "sigla_uf"], how="left")
df_pbf_2024 = df_pbf_2024.merge(df_ideb, on=["municipio", "sigla_uf"], how="left")
```

```python
df_pbf_2023["ano"] = 2023
df_pbf_2023 = df_pbf_2023.filter(items=["ano", "municipio", "sigla_uf", "proporcao_pbf_municipio", "ideb_medio", "tendencia_ideb","gini_uf"]).rename(columns={"ideb_medio": "ideb_medio_municipio", "tendencia_ideb": "tendencia_ideb_municipio"})
```

```python
df_pbf_2024["ano"] = 2024
df_pbf_2024 = df_pbf_2024.filter(items=["ano", "municipio", "sigla_uf", "proporcao_pbf_municipio", "ideb_medio", "tendencia_ideb","gini_uf"]).rename(columns={"ideb_medio": "ideb_medio_municipio", "tendencia_ideb": "tendencia_ideb_municipio"})
```

```python
df_socio_econ = pd.concat([df_pbf_2023, df_pbf_2024])
```
