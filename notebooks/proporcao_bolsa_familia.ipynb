## Estruturação de arquivos CSV com a proporção de população inscrita no Bolsa Família nos anos de 2023 e 2024, para cada município brasileiro.

```python
import pandas as pd
import numpy as np
```

```python
df_bolsa_familia_2023 = pd.read_csv("/content/bolsa_familia_2023.csv", delimiter=';')
df_bolsa_familia_2024 = pd.read_csv("/content/bolsa_familia_2024.csv", delimiter=';')
df_populacao = pd.read_csv("/content/br_ibge_populacao_municipio.csv")
df_city_code = pd.read_csv("/content/br_bd_diretorios_brasil_municipio.csv")
```

```python
df_bolsa_familia_2023.head()
```

```python
df_bolsa_familia_2023 = df_bolsa_familia_2023.filter(items=["Unidade Territorial", "UF", "Pessoas beneficiárias"]).rename(columns={"Unidade Territorial": "municipio", "UF": "sigla_uf"})
df_bolsa_familia_2023['municipio'] = df_bolsa_familia_2023['municipio'].str.upper()
df_bolsa_familia_2023.head()
```

```python
df_bolsa_familia_2024.head()
```

```python
df_bolsa_familia_2024 = df_bolsa_familia_2024.filter(items=["Unidade Territorial", "UF", "Pessoas beneficiárias no PBF (a partir de Mar/2023)"]).rename(columns={"Unidade Territorial": "municipio", "UF": "sigla_uf", "Pessoas beneficiárias no PBF (a partir de Mar/2023)": "Pessoas beneficiárias"})
df_bolsa_familia_2024['municipio'] = df_bolsa_familia_2024['municipio'].str.upper()
df_bolsa_familia_2024.head()
```

```python
df_populacao_2023 = df_populacao[df_populacao['ano'] == 2023]
df_populacao_2023.head()
```

```python
df_populacao_2024 = df_populacao[df_populacao['ano'] == 2024]
df_populacao_2024.head()
```

```python
df_city_code['nome'] = df_city_code['nome'].str.upper()
df_city_code.head()
```

```python
df_populacao_2023 = df_populacao_2023.merge(df_city_code, how='left', left_on='id_municipio', right_on='id_municipio').filter(items=['nome', 'sigla_uf_x', 'populacao']).rename(columns={'sigla_uf_x': 'sigla_uf', 'nome':'municipio'})
```

```python
df_populacao_2024 = df_populacao_2024.merge(df_city_code, how='left', left_on='id_municipio', right_on='id_municipio').filter(items=['nome', 'sigla_uf_x', 'populacao']).rename(columns={'sigla_uf_x': 'sigla_uf', 'nome':'municipio'})
```

```python
df_bolsa_familia_2023.isna().sum()
```

```python
df_bolsa_familia_2024.isna().sum()
```

```python
df_populacao_2023.isna().sum()
```

```python
df_populacao_2024.isna().sum()
```

```python
proporcao_2023 = df_bolsa_familia_2023.merge(df_populacao_2023, how='left', on=['municipio', 'sigla_uf'])
proporcao_2023['proporcao'] = (proporcao_2023['Pessoas beneficiárias'] / proporcao_2023['populacao']).round(2)
proporcao_2023.head()
```

```python
proporcao_2023[proporcao_2023['populacao'].isna()]
```

```python
proporcao_2023.to_csv("br_proporcao_pbf_2023.csv")
```

```python
proporcao_2024 = df_bolsa_familia_2024.merge(df_populacao_2024, how='left', on=['municipio', 'sigla_uf'])
proporcao_2024["proporcao"] = (proporcao_2024["Pessoas beneficiárias"] / proporcao_2024["populacao"]).round(2)
proporcao_2024.head()
```

```python
proporcao_2024[proporcao_2024['populacao'].isna()]
```

```python
proporcao_2024.to_csv("br_proporcao_pbf_2024.csv")
```
