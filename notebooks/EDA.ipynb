Análise de alfabetização por região

```python
df["regiao"] = df['sigla_uf'].map(regioes_br)
```

```python
proporcao_alfabetizados_regiao = ((df.groupby("regiao")["target_alfabetizado"].sum().reset_index().set_index("regiao")['target_alfabetizado']) / 
                           (df.groupby("regiao")["target_alfabetizado"].count().reset_index().set_index("regiao")['target_alfabetizado'])).reset_index()
```

```python
proporcao_alfabetizados_regiao = proporcao_alfabetizados_regiao.rename(columns={"target_alfabetizado": "proporcao_alfabetizados"})
```

```python
proporcao_alfabetizados_regiao.sort_values(by="proporcao_alfabetizados", ascending=False)
```

O índice de Gini é um indicador matemático criado para medir o grau de desigualdade na distribuição de renda ou de riqueza em uma população.

Fica em uma escala de 0 a 1, onde zero indica igualdade perfeita (todo mundo ganha exatamente o mesmo valor) e um indica desigualdade máxima (uma única pessoa detém toda a renda do lugar e os outros nada).

```python
gini_medio_regiao = df.groupby('regiao')['feat_gini_uf'].mean().reset_index().sort_values(by="feat_gini_uf", ascending=False)

plt.figure(figsize=(10, 6))

fig, ax = plt.subplots()
bars = ax.bar(gini_medio_regiao['regiao'], gini_medio_regiao['feat_gini_uf'].round(3), color='mediumslateblue')
ax.bar_label(bars, padding=3)

plt.ylim(0, 1)

plt.title('Índice Gini Médio por Região')
plt.xlabel('Região')
plt.ylabel('Índice Gini')
plt.show()
```

Nota-se uma tendência que as regiões com maior proporção de alfabetizados sejam aquelas com índice de Gini mais baixos (menor desigualdade na distribuição de renda).
